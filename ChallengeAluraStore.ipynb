### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
facturacion = {
    "Tienda 1": (tienda["Precio"] * tienda["Cantidad de cuotas"]).sum(),
    "Tienda 2": (tienda2["Precio"] * tienda2["Cantidad de cuotas"]).sum(),
    "Tienda 3": (tienda3["Precio"] * tienda3["Cantidad de cuotas"]).sum(),
    "Tienda 4": (tienda4["Precio"] * tienda4["Cantidad de cuotas"]).sum()
}

pd.Series(facturacion, name="Facturación Total")
```

# 2. Ventas por categoría

```python
categorias_populares = {
    "Tienda 1": tienda["Categoría del Producto"].value_counts().idxmax(),
    "Tienda 2": tienda2["Categoría del Producto"].value_counts().idxmax(),
    "Tienda 3": tienda3["Categoría del Producto"].value_counts().idxmax(),
    "Tienda 4": tienda4["Categoría del Producto"].value_counts().idxmax()
}

pd.Series(categorias_populares, name="Categoría Más Popular")
```

# 3. Calificación promedio de la tienda


```python
evaluacion_promedio = {
    "Tienda 1": tienda["Calificación"].mean(),
    "Tienda 2": tienda2["Calificación"].mean(),
    "Tienda 3": tienda3["Calificación"].mean(),
    "Tienda 4": tienda4["Calificación"].mean()
}

pd.Series(evaluacion_promedio, name="Promedio Evaluación")
```

# 4. Productos más y menos vendidos

```python
productos_vendidos = {}

for i, df in enumerate([tienda, tienda2, tienda3, tienda4], 1):
    ventas = df.groupby("Producto")["Cantidad de cuotas"].sum()  # Usar 'Producto' y 'Cantidad de cuotas'
    productos_vendidos[f"Tienda {i} - Más vendido"] = ventas.idxmax()
    productos_vendidos[f"Tienda {i} - Menos vendido"] = ventas.idxmin()

pd.Series(productos_vendidos)
```

# 5. Envío promedio por tienda

```python
envio_promedio = {
    "Tienda 1": tienda["Costo de envío"].mean(),
    "Tienda 2": tienda2["Costo de envío"].mean(),
    "Tienda 3": tienda3["Costo de envío"].mean(),
    "Tienda 4": tienda4["Costo de envío"].mean()
}

pd.Series(envio_promedio, name="Promedio Envío")
```
